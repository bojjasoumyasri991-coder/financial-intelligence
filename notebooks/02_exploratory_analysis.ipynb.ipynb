```python
numeric_cols = [
    "sales",
    "expenses",
    "operating_profit",
    "opm_percentage",
    "other_income",
    "interest",
    "depreciation",
    "profit_before_tax",
    "tax_percentage",
    "net_profit"
]

for col in numeric_cols:
    profit_loss[col] = pd.to_numeric(
        profit_loss[col],
        errors="coerce"
    )
```

```python
profit_loss["net_margin"] = (
    profit_loss["net_profit"]
    / profit_loss["sales"]
) * 100
```

```python
profit_loss["operating_margin"] = (
    profit_loss["operating_profit"]
    / profit_loss["sales"]
) * 100
```

```python
profit_loss["interest_burden"] = (
    profit_loss["interest"]
    / profit_loss["sales"]
) * 100
```

```python
top_profit = (
    profit_loss.groupby("company_id")["net_profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_profit
```

```python
sales_growth = (
    profit_loss.groupby("company_id")["sales"]
    .agg(["first", "last"])
)

sales_growth["sales_cagr"] = (
    (
        sales_growth["last"]
        / sales_growth["first"]
    ) ** (1/10) - 1
) * 100

sales_growth.sort_values(
    "sales_cagr",
    ascending=False
).head(10)
```

```python
balance_sheet = pd.read_excel(
    "data/balancesheet.xlsx",
    skiprows=1
)

balance_sheet.head()
```

```python
balance_sheet["debt_to_equity"] = (
    balance_sheet["borrowings"]
    /
    (
        balance_sheet["equity_capital"]
        + balance_sheet["reserves"]
    )
)

balance_sheet[
    [
        "company_id",
        "year",
        "borrowings",
        "equity_capital",
        "reserves",
        "debt_to_equity"
    ]
].head()
```

```python
balance_sheet["asset_turnover"] = (
    balance_sheet["total_assets"]
    /
    (
        balance_sheet["equity_capital"]
        + balance_sheet["reserves"]
    )
)
```

```python
balance_sheet[
    [
        "company_id",
        "year",
        "asset_turnover"
    ]
].head()
```

```python
financial_df = pd.merge(
    profit_loss,
    balance_sheet,
    on=["company_id", "year"],
    how="inner"
)

financial_df.head()
```

```python
financial_df["roe"] = (
    financial_df["net_profit"]
    /
    (
        financial_df["equity_capital"]
        + financial_df["reserves"]
    )
) * 100
```

```python
financial_df["roa"] = (
    financial_df["net_profit"]
    /
    financial_df["total_assets"]
) * 100
```

```python
financial_df["interest_coverage"] = (
    financial_df["operating_profit"]
    /
    financial_df["interest"]
)
```

```python
top_roe = (
    financial_df.groupby("company_id")["roe"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

top_roe
```

```python
top_roe.plot(
    kind="bar",
    figsize=(12,6),
    title="Top ROE Companies",
    color="orange"
)

plt.show()
```

```python
financial_df["roe"] = financial_df["roe"].replace(
    [np.inf, -np.inf],
    np.nan
)
```

```python
financial_df = financial_df[
    financial_df["roe"] < 100
]
```

```python
top_roe = (
    financial_df.groupby("company_id")["roe"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

top_roe
```

```python
top_roe.plot(
    kind="bar",
    figsize=(12,6),
    title="Top ROE Companies (Cleaned)",
    color="orange"
)

plt.show()
```

```python
company_metrics = financial_df.groupby("company_id").agg({
    "roe": "mean",
    "net_margin": "mean",
    "roa": "mean",
    "debt_to_equity": "mean",
    "interest_coverage": "mean"
}).reset_index()

company_metrics.head()
```

```python
company_metrics["health_score"] = (
    company_metrics["roe"] * 0.30
    +
    company_metrics["net_margin"] * 0.25
    +
    company_metrics["roa"] * 0.20
    +
    company_metrics["interest_coverage"] * 0.15
    -
    company_metrics["debt_to_equity"] * 0.10
)
```

```python
company_metrics = company_metrics.sort_values(
    "health_score",
    ascending=False
)

company_metrics.head(10)
```

```python
top_health = company_metrics.head(10)

top_health.plot(
    x="company_id",
    y="health_score",
    kind="bar",
    figsize=(12,6),
    title="Top Financially Healthy Companies",
    color="green"
)

plt.show()
```
